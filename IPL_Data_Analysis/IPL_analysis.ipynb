```python
#Replace "NA" with NULL
from pyspark.sql.functions import col, when, year
silver_matches_df = bronze_matches_df

string_cols = [
    "city",
    "player_of_match",
    "winner",
    "method"
]

for c in string_cols:
    silver_matches_df = silver_matches_df.withColumn(c,
                                                     when(col(c) == "NA", None)
                                                     .otherwise(col(c)))
```

```python
# Create season_clean
silver_matches_df = silver_matches_df.withColumn('season_clean',
                                                 when(
                                                     col("season").isNull(),
                                                     year(col("date"))
                                                     ).otherwise(col("season")))
```

```python
# Create match_outcome
silver_matches_df = silver_matches_df.withColumn(
    "match_outcome",
    when(col("result").isin("runs", "wickets"), "Won")
    .when(col("result") == "tie", "Tie")
    .when(col("result") == "no result", "No Result")
    .otherwise("Unknown"))
```

### Gold Batting Statistics
- Player-level batting statistics.

```python
from pyspark.sql.functions import col,count,sum,avg,when,round,countDistinct

gold_batting_stats_df = (gold_delivery_match_df.groupBy('batter')
                                               .agg(
                                                   # total matches played by batsman
                                                   countDistinct("match_id").alias("matches_played"),

                                                   # total runs by batsman
                                                   sum('batsman_runs').alias('total_runs'),

                                                   # calculate total balls faced by batsman
                                                   sum(
                                                       when(col('is_legal_delivery') == True, 1)
                                                       .otherwise(0)
                                                   ).alias('balls_faced'),

                                                   #Total fours by batsman
                                                   sum(
                                                       when(col('is_four') == True,1)
                                                        .otherwise(0)
                                                   ).alias('fours'),

                                                   #Total sixes by batsman
                                                   sum(
                                                        when(col("is_six") == True, 1)
                                                        .otherwise(0)
                                                    ).alias("sixes"),
                                               
                                               #Total dismissals
                                               sum(
                                                    when(
                                                        (col("is_wicket") == 1) &
                                                        (col("player_dismissed") == col("batter")),
                                                        1
                                                    ).otherwise(0)
                                                ).alias("dismissals")
                                                   
                                                   
                                               )
                         )
```

```python
gold_batting_stats_df.show(10)
```

```python
# calculate strike rate for each player
from pyspark.sql.functions import when, col, round

gold_batting_stats_df = gold_batting_stats_df.withColumn(
    "strike_rate",
    when(
        col("balls_faced") > 0,
        round(
            col("total_runs") / col("balls_faced") * 100,
            2
        )
    ).otherwise(0)
)
```

### Gold bowling Statistics
- Player-level bowling statistics.

```python
from pyspark.sql.functions import (
    col,
    sum,
    when,
    countDistinct,
    round
)

gold_bowling_stats_df = (
    gold_delivery_match_df
    .groupBy("bowler")
    .agg(

        # Distinct matches bowled
        countDistinct("match_id").alias("matches_played"),

        # Legal deliveries
        sum(
            when(
                col("is_legal_delivery") == True,
                1
            ).otherwise(0)
        ).alias("balls_bowled"),

        # Runs conceded by bowler
        sum(
            col("batsman_runs") +
            when(
                col("extras_type").isin("wides", "noballs"),
                col("extra_runs")
            ).otherwise(0)
        ).alias("runs_conceded"),

        # Bowler credited wickets
        sum(
            when(
                (col("is_wicket") == 1) &
                (
                    col("dismissal_kind").isin(
                        "caught",
                        "bowled",
                        "lbw",
                        "stumped",
                        "caught and bowled",
                        "hit wicket"
                    )
                ),
                1
            ).otherwise(0)
        ).alias("wickets"),

        # Fours conceded
        sum(
            when(
                (col("batsman_runs") == 4) &
                (col("is_boundary") == True),
                1
            ).otherwise(0)
        ).alias("fours_conceded"),

        # Sixes conceded
        sum(
            when(
                (col("batsman_runs") == 6) &
                (col("is_boundary") == True),
                1
            ).otherwise(0)
        ).alias("sixes_conceded")
    )
)

# Economy rate
gold_bowling_stats_df = gold_bowling_stats_df.withColumn(
    "economy",
    when(
        col("balls_bowled") > 0,
        round(
            col("runs_conceded") / col("balls_bowled") * 6,
            2
        )
    ).otherwise(0)
)
```

### Gold team Statistics
- team performance

```python
gold_delivery_match_df.printSchema()
```

```python
from pyspark.sql.functions import (
    col,
    sum,
    when,
    count,
    countDistinct,
    round,
    lit
)

# ---------------------------------------------------------
# 1. Create team-season match participation
# ---------------------------------------------------------

team1_df = (
    silver_matches_df
    .select(
        col("season_clean").alias("season"),
        col("id").alias("match_id"),
        col("team1").alias("team"),
        col("winner"),
        col("match_outcome")
    )
)

team2_df = (
    silver_matches_df
    .select(
        col("season_clean").alias("season"),
        col("id").alias("match_id"),
        col("team2").alias("team"),
        col("winner"),
        col("match_outcome")
    )
)

team_matches_df = team1_df.unionByName(team2_df)


# ---------------------------------------------------------
# 2. Standardize historical team names
# ---------------------------------------------------------

team_matches_df = team_matches_df.withColumn(
    "team_standardized",
    when(
        col("team") == "Delhi Daredevils",
        "Delhi Capitals"
    ).when(
        col("team") == "Kings XI Punjab",
        "Punjab Kings"
    ).when(
        col("team") == "Royal Challengers Bangalore",
        "Royal Challengers Bengaluru"
    ).otherwise(
        col("team")
    )
)


# ---------------------------------------------------------
# 3. Match-level team statistics
# ---------------------------------------------------------

team_match_stats_df = (
    team_matches_df
    .groupBy(
        "season",
        "team",
        "team_standardized"
    )
    .agg(

        countDistinct("match_id").alias("matches_played"),

        # Team actually won the match
        sum(
            when(
                (col("match_outcome") == "Won") &
                (col("winner") == col("team")),
                1
            ).otherwise(0)
        ).alias("matches_won"),

        # Match had a winner, but it was the opponent
        sum(
            when(
                (col("match_outcome") == "Won") &
                (col("winner") != col("team")),
                1
            ).otherwise(0)
        ).alias("matches_lost"),

        # Match ended in a tie
        sum(
            when(
                col("match_outcome") == "Tie",
                1
            ).otherwise(0)
        ).alias("ties"),

        # Match had no result
        sum(
            when(
                col("match_outcome") == "No Result",
                1
            ).otherwise(0)
        ).alias("no_results")
    )
)

# ---------------------------------------------------------
# 4. Team batting statistics
# ---------------------------------------------------------

team_batting_df = (
    gold_delivery_match_df
    .groupBy(
        col("season_clean"),
        col("batting_team")
    )
    .agg(
        sum("batsman_runs").alias("total_runs"),

        sum(
            when(
                col("is_wicket") == 1,
                1
            ).otherwise(0)
        ).alias("wickets_lost")
    )
    .withColumnRenamed("season_clean", "season")
    .withColumnRenamed("batting_team", "team")
)


# ---------------------------------------------------------
# 5. Join match statistics + batting statistics
# ---------------------------------------------------------

gold_team_performance_df = (
    team_match_stats_df
    .join(
        team_batting_df,
        on=["season", "team"],
        how="left"
    )
    .withColumn(
        "win_percentage",
        when(
            col("matches_played") > 0,
            round(
                col("matches_won") /
                col("matches_played") * 100,
                2
            )
        ).otherwise(0)
    )
)


# ---------------------------------------------------------
# 6. Select final columns
# ---------------------------------------------------------

gold_team_performance_df = gold_team_performance_df.select(
    "season",
    "team",
    "team_standardized",
    "matches_played",
    "matches_won",
    "matches_lost",
    "ties",
    "no_results",
    "win_percentage",
    "total_runs",
    "wickets_lost"
)
```
