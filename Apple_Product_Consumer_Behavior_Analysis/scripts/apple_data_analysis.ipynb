```python
class FirstWorkFlow:
    '''
    ETL piepline to retrieve all the customer who bought Airpods just after buying iPhone and load the data to different sink (parquet file) 
    file path to upload result \ output - /Volumes/upskill/pyspark_learning/apple_analysis_output
    '''
    def __init__(self):
        pass

    def runner(self):

        # customer who bought airpods after buying Iphone

        #step1 : Extract all required dataframes
        inputDFs = read_raw_data().extract()

         #step2 : transform the dataframes
        firstTransformdf = FirstTranformer().transform(inputDFs)

        #step3 : Load all required data to different sink
        AirPodsAfterIphoneLoader(firstTransformdf).sink()
        #firstTransform.show()
```

```python
 
'''
    ETL piepline to retrieve List all the products bought by customer after their initial purchase of a product.
    dataset used : 

 inputDFs = {
            "transactionInputdf": transactionInputdf,
            "customerInputdf" :customerInputdf,
            "productInputdf":productInputdf
        }
'''

class AllProductsExceptInitial:

    def __init__(self):
        pass

    def runner(self):

        # customer who bought airpods after buying Iphone

        #step1 : Extract all required dataframes
        inputDFs = read_raw_data().extract()

         #step2 : transform the dataframes to get products bought by each customer except first product 
        allprodsexceptinitialTransformdf = AllProductsExceptInitialTranformer().transform(inputDFs)

        #step3 : Load all required data to different sink
        AllprodsexceptinitialLoader(allprodsexceptinitialTransformdf).sink()
```

```python

'''
    ETL piepline to Determine the time between buying an iphone and buying airpods for each customer.
    dataset used : 

 inputDFs = {
            "transactionInputdf": transactionInputdf,
            "customerInputdf" :customerInputdf,
            "productInputdf":productInputdf
        }
'''

class TimeBetweenIphoneAndAirPods:

    def __init__(self):
        pass

    def runner(self):

        # customer who bought airpods after buying Iphone

        #step1 : Extract all required dataframes
        inputDFs = read_raw_data().extract()

        #step2 : transform the dataframes to get products bought by each customer except first product 
        avgLadTimebuyingIphoneAndAirPodsDf = TimeBetweenIphoneAndAirPodsTranformer().transform(inputDFs)

        #step3 : Load all required data to different sink
        AvgLagTimebuyingIphoneAndAirPodsLoader(avgLadTimebuyingIphoneAndAirPodsDf).sink()
```

```python
# Identify the top 3 selling products in each category by total revenue
class Top3PerCategory:

    def __init__(self):
        pass

    def runner(self):

        #step1 : Extract all required dataframes
        inputDFs = read_raw_data().extract()

        #step2 : transform the dataframes to get products bought by each customer except first product 
        Top3PerCategoryDf = Top3PerCategoryTranformer().transform(inputDFs)

        #step3 : Load all required data to different sink
        AvgLagTimebuyingIphoneAndAirPodsLoader(Top3PerCategoryDf).sink()

```

```python
class WorkFlowRunner:

    def __init__(self, name):
        self.name = name

    def runner(self):
        if self.name == "firstWorkFlow":
            return FirstWorkFlow().runner()
        elif self.name == "secondWorkFlow":
            return SecondWorkFlow().runner()
        elif self.name =="allproductsexceptinitial":
            return AllProductsExceptInitial().runner()
        elif self.name =="timeBetweenIphoneAndAirPods":
            return TimeBetweenIphoneAndAirPods().runner()
        elif self.name =="top3_per_category":
            return Top3PerCategory().runner()
        else:
            raise ValueError(f"Not Implemented for {self.name}")

name = "top3_per_category" 

workFlowrunner = WorkFlowRunner(name).runner()
```
